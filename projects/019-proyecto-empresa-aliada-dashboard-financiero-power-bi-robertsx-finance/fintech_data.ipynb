# Análisis y Validación de Datos Financieros

## RobertsX Finance - Business Intelligence Project

Durante este notebook realicé la exploración inicial y validación de los datasets financieros generados mediante Python para alimentar el dashboard desarrollado en Power BI.

El objetivo fue revisar la estructura de la información, validar consistencia entre tablas y preparar la base analítica utilizada posteriormente en el modelo de Business Intelligence.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option('display.max_columns', None)
```

## Carga de información

Los archivos utilizados en este análisis fueron generados mediante el script `generate_fintech_data.py`.

Los datasets contienen información simulada de clientes, productos financieros, países y transacciones.

Los archivos CSV generados no forman parte del repositorio debido a su tamaño. La generación completa puede reproducirse ejecutando nuevamente el script principal.

```python
customers = pd.read_csv('../data/customers.csv')
transactions = pd.read_csv('../data/transactions.csv')
products = pd.read_csv('../data/products.csv')
countries = pd.read_csv('../data/countries.csv')
```

## Exploración inicial

Realicé una revisión general para conocer dimensiones, estructura y características principales de cada fuente.

```python
datasets = {
    'Clientes': customers,
    'Transacciones': transactions,
    'Productos': products,
    'Países': countries
}

for name, df in datasets.items():
    print('\n', name)
    print('Registros:', df.shape[0])
    print('Columnas:', df.shape[1])
```

## Validación de calidad

Realicé una revisión de valores faltantes y duplicados para verificar la consistencia inicial de la información.

```python
for name, df in datasets.items():
    print('\nDataset:', name)
    print('Valores nulos:')
    print(df.isnull().sum())
    print('Duplicados:', df.duplicated().sum())
```

## Indicadores principales

Construí métricas iniciales para validar el comportamiento general de la información financiera.

```python
total_customers = customers['customer_id'].nunique()
total_transactions = transactions['transaction_id'].nunique()
total_revenue = transactions['revenue_usd'].sum()

print('Clientes:', total_customers)
print('Transacciones:', total_transactions)
print('Ingresos generados:', round(total_revenue,2))
```

```python
transactions.groupby('transaction_type')['revenue_usd'].sum().sort_values().plot(kind='bar', figsize=(10,5))
plt.title('Ingresos por tipo de transacción')
plt.xlabel('Tipo de operación')
plt.ylabel('Ingresos USD')
plt.xticks(rotation=45)
plt.show()
```

## Conclusión

El análisis permitió validar que la información generada cuenta con una estructura adecuada para ser utilizada dentro de una solución de Business Intelligence.

Después de revisar consistencia, dimensiones y métricas principales, los datasets fueron utilizados como fuente para la construcción del dashboard financiero desarrollado en Power BI.
