```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
from sklearn.preprocessing import MinMaxScaler
import warnings
warnings.filterwarnings('ignore')
print('Librerías cargadas correctamente')
```

## Carga de Datos

Por motivos de peso en GitHub, los datasets no se incluyen en el repositorio.
Se documenta la estructura utilizada durante el desarrollo.

```python
data = {
    'YEAR':[1990,1995,2000,2005,2010,2015,2020],
    'GDP':[500,650,900,1200,1500,1800,2000]
}
df = pd.DataFrame(data)
df.head()
```

## Exploración Inicial

```python
df.info()
df.describe()
```

## Preparación de Variables

```python
X = df['YEAR'].values
y = df['GDP'].values
```

## Función Logística

```python
def logistic_function(x, L, k, x0):
    return L / (1 + np.exp(-k*(x-x0)))
```

## Ajuste Inicial

```python
params_original, _ = curve_fit(logistic_function, X, y, maxfev=10000)
params_original
```

## Normalización

```python
scaler = MinMaxScaler()
y_scaled = scaler.fit_transform(y.reshape(-1,1)).flatten()
```

## Ajuste Normalizado

```python
params_scaled, _ = curve_fit(logistic_function, X, y_scaled, maxfev=10000)
params_scaled
```

## Reconstrucción del Modelo

```python
y_pred_scaled = logistic_function(X, *params_scaled)
y_pred_original = scaler.inverse_transform(y_pred_scaled.reshape(-1,1)).flatten()
```

## Visualización Final

```python
plt.scatter(X, y)
plt.plot(X, y_pred_original, color='red')
plt.show()
```

## Evaluación del Modelo

```python
error = y - y_pred_original
print('MSE:', np.mean(error**2))
print('MAE:', np.mean(np.abs(error)))
```
