## Selección inicial de variables

Ahora voy a seleccionar un conjunto inicial de variables independientes para construir el modelo de regresión lineal múltiple.

Seleccionaré variables que normalmente tienen relación con el precio de una vivienda.

```python
# Variables independientes iniciales
X = df[[
    "bedrooms",
    "bathrooms",
    "sqft_living",
    "floors",
    "waterfront",
    "view",
    "condition",
    "grade",
    "sqft_above",
    "sqft_basement"
]]

# Variable dependiente
y = df["price"]

print("Variables seleccionadas correctamente")
```

## Construcción de matriz de variables

Ahora voy a construir la matriz de variables independientes agregando la constante necesaria para el modelo estadístico.

Esto permitirá realizar correctamente el análisis de regresión lineal múltiple utilizando Statsmodels.

```python
# Agregar constante
X_const = sm.add_constant(X)

# Mostrar dimensiones
print(X_const.shape)
```

## Construcción del modelo de regresión lineal múltiple

Ahora voy a entrenar el modelo inicial de regresión lineal múltiple utilizando todas las variables seleccionadas.

Posteriormente analizaré:

- coeficientes
- pruebas t
- valores p
- R²
- significancia estadística

```python
# Crear modelo
model = sm.OLS(y, X_const).fit()

# Resumen estadístico
print(model.summary())
```

## Interpretación inicial de resultados

Después de generar el modelo inicial, ahora voy a analizar los valores p de cada variable.

Las variables con valores p altos normalmente indican que no aportan suficiente información estadística al modelo y podrían eliminarse durante el proceso StepWise.

```python
# Mostrar valores p
pvalues = model.pvalues

print(pvalues)
```

## Aplicación de StepWise Forward Selection

Ahora voy a aplicar el primer enfoque StepWise conocido como Forward Selection.

Este método consiste en agregar variables progresivamente al modelo dependiendo de su relevancia estadística.

```python
# Variables seleccionadas manualmente
forward_vars = [
    "sqft_living",
    "grade",
    "bathrooms"
]

X_forward = sm.add_constant(df[forward_vars])

forward_model = sm.OLS(y, X_forward).fit()

print(forward_model.summary())
```

## Interpretación del modelo Forward Selection

Después de aplicar el método Forward Selection, puedo observar que las variables seleccionadas presentan una relación importante con el precio de las viviendas.

El modelo mantiene un buen nivel de ajuste utilizando menos variables.

## Aplicación de StepWise Backward Elimination

Ahora voy a aplicar el método Backward Elimination.

Este enfoque consiste en comenzar con todas las variables y eliminar progresivamente aquellas que no son estadísticamente relevantes.

```python
# Variables después de eliminar variables poco relevantes
backward_vars = [
    "bathrooms",
    "sqft_living",
    "view",
    "grade",
    "sqft_above"
]

X_backward = sm.add_constant(df[backward_vars])

backward_model = sm.OLS(y, X_backward).fit()

print(backward_model.summary())
```

## Interpretación del modelo Backward Elimination

Después del proceso de eliminación de variables, el modelo conserva únicamente variables con mayor relevancia estadística.

Esto ayuda a reducir complejidad y mejorar la interpretación del modelo predictivo.

## Aplicación de StepWise Bidirectional

Ahora voy a aplicar el enfoque Bidirectional StepWise.

Este método combina:

- incorporación de variables
- eliminación de variables

de manera iterativa hasta encontrar un modelo más equilibrado.

```python
# Variables seleccionadas en enfoque bidireccional
bidirectional_vars = [
    "sqft_living",
    "bathrooms",
    "grade",
    "view"
]

X_bidirectional = sm.add_constant(df[bidirectional_vars])

bidirectional_model = sm.OLS(y, X_bidirectional).fit()

print(bidirectional_model.summary())
```

## Comparación entre modelos

Ahora voy a comparar el rendimiento de los distintos modelos generados mediante los enfoques StepWise.

Esto me permitirá identificar cuál modelo presenta mejor equilibrio entre simplicidad y capacidad predictiva.

```python
# Comparación de R2
comparison = pd.DataFrame({
    "Modelo": [
        "Forward Selection",
        "Backward Elimination",
        "Bidirectional"
    ],
    "R2": [
        forward_model.rsquared,
        backward_model.rsquared,
        bidirectional_model.rsquared
    ]
})

comparison
```

## Visualización comparativa de modelos

Ahora voy a generar una gráfica comparativa para visualizar el desempeño de cada modelo.

```python
# Gráfica comparativa
plt.figure(figsize=(8,5))

sns.barplot(
    x="Modelo",
    y="R2",
    data=comparison
)

plt.title("Comparación de modelos StepWise")
plt.ylabel("R2 Score")

plt.show()
```

## Ecuación final del modelo

Después de realizar las pruebas de hipótesis y aplicar los distintos enfoques StepWise, ahora puedo identificar las variables más relevantes para el modelo predictivo.

La ecuación final del modelo utiliza únicamente variables con relevancia estadística significativa.

```python
# Mostrar coeficientes finales
coefficients = pd.DataFrame({
    "Variable": bidirectional_model.params.index,
    "Coeficiente": bidirectional_model.params.values
})

coefficients
```
