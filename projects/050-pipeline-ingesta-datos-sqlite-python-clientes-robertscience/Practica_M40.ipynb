```python
import sqlite3
import pandas as pd
import os
```

## 📥 Carga de datos

Se realiza la lectura del archivo `clientes.csv` utilizando Pandas.  
Este dataset contiene información de clientes como nombre, dirección, país y compras realizadas.

Este paso permite preparar los datos para su posterior almacenamiento en SQLite.

```python
df = pd.read_csv(r"D:\Documentos\Ebac\Finales\Práctica M40\clientes.csv")
df.head()
```

## 🔎 Vista preliminar de los datos

Se muestran los primeros registros del dataset para validar su estructura, tipos de datos y consistencia antes de realizar operaciones de transformación o carga en la base de datos.

## ✂️ Selección de muestra de datos

Se seleccionan únicamente los primeros 5 registros del dataset.  
Esto se realiza para trabajar con una muestra controlada que será insertada en la base de datos SQLite.

```python
df_5 = df.head(5)
df_5
```

## 🗄️ Creación de base de datos SQLite

Se establece una conexión con una base de datos local llamada `Base.db`.

Si el archivo no existe, SQLite lo crea automáticamente en el directorio del proyecto.

```python
conn = sqlite3.connect("Base.db")
cursor = conn.cursor()
```

## 🧱 Creación de tabla Clientes

Se define la estructura de la tabla `Clientes`, la cual almacenará la información del dataset.

Se incluyen campos como:
- ID del cliente
- Nombre
- Contacto
- Dirección
- Ciudad
- Código postal
- País

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS Clientes (
    CustomerID INTEGER,
    CustomerName TEXT,
    ContactName TEXT,
    Address TEXT,
    City TEXT,
    PostalCode TEXT,
    Country TEXT
)
""")

conn.commit()
```

## 📤 Inserción de datos en la base de datos

Se insertan los 5 registros seleccionados del dataset dentro de la tabla `Clientes`.

Este proceso utiliza un ciclo para recorrer cada fila y ejecutar una sentencia SQL INSERT.

```python
# Limpia tabla antes de insertar (evita duplicados en notebooks)
cursor.execute("DELETE FROM Clientes")
conn.commit()

# Inserta solo 5 registros
for _, row in df.head(5).iterrows():
    cursor.execute("""
    INSERT INTO Clientes VALUES (?, ?, ?, ?, ?, ?, ?)
    """, (
        row["CustomerID"],
        row["CustomerName"],
        row["ContactName"],
        row["Address"],
        row["City"],
        row["PostalCode"],
        row["Country"]
    ))

conn.commit()
```

## 🔍 Verificación de estructura de base de datos

Se consulta el sistema interno de SQLite para validar que la tabla `Clientes` fue creada correctamente.

```python
cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
print(cursor.fetchall())
```

## 📊 Consulta final de validación

Se ejecuta una consulta SQL para recuperar todos los registros almacenados en la tabla `Clientes`.

Esto confirma que la inserción de datos fue exitosa.

```python
df_result = pd.read_sql_query("SELECT * FROM Clientes", conn)
df_result
```

“El flujo demuestra la integración básica entre ingesta, almacenamiento y consulta de datos estructurados.”

```python
conn.close()
```
