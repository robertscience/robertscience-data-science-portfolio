## Importación de librerías

Primero voy a importar las librerías necesarias para trabajar con series de tiempo, análisis estadístico, visualización y descarga de datos financieros.

Estas herramientas permitirán realizar el análisis exploratorio, pruebas estadísticas y pronósticos solicitados en esta práctica.

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import yfinance as yf

from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.tsa.stattools import adfuller
from statsmodels.graphics.tsaplots import plot_acf

# Configuración visual
plt.style.use("ggplot")

print("Librerías cargadas correctamente")
```

## Descarga de datos históricos desde Yahoo Finance

Ahora voy a descargar la información histórica diaria de precios de las acciones de IBM y Walmart utilizando Yahoo Finance.

La práctica solicita trabajar con información correspondiente a los últimos 5 años para analizar el comportamiento temporal de ambas series financieras.

```python
# Descargar datos históricos (últimos 5 años)
ibm = yf.download("IBM", period="5y")
wmt = yf.download("WMT", period="5y")

print("Datos descargados correctamente")
```

## Visualización inicial de los datasets

Después de descargar la información histórica, ahora voy a revisar las primeras filas de ambos datasets para validar que los datos fueron cargados correctamente.

Esto permite observar la estructura general de cada conjunto de datos y confirmar que las variables financieras están disponibles para el análisis.

```python
print("IBM")
display(ibm.head())

print("Walmart")
display(wmt.head())
```

## Información general de los datasets

Ahora voy a revisar la estructura general de ambos datasets para conocer:

- número de registros
- nombres de columnas
- tipos de datos
- posibles valores faltantes

Esta validación es importante antes de iniciar el análisis de series de tiempo.

```python
print("IBM")
ibm.info()

print("\nWalmart")
wmt.info()
```

## Selección de precios de cierre

Para el análisis de series de tiempo utilizaré la variable **Close**, ya que representa el precio de cierre diario de cada acción.

Esta variable es una de las más utilizadas en análisis financiero para estudiar tendencias, correlaciones y realizar pronósticos.

```python
ibm_close = ibm["Close"]
wmt_close = wmt["Close"]

print("Series de precios de cierre preparadas correctamente")
```

## Gráfico de comportamiento histórico de precios

Ahora voy a visualizar el comportamiento temporal de los precios de cierre de IBM y Walmart durante los últimos 5 años.

Esta gráfica permitirá observar tendencias generales, cambios en volatilidad y comportamiento financiero de ambas acciones.

```python
plt.figure(figsize=(12,6))

plt.plot(ibm_close, label="IBM")
plt.plot(wmt_close, label="Walmart")

plt.title("Precios de cierre históricos")
plt.xlabel("Fecha")
plt.ylabel("Precio")
plt.legend()

plt.show()
```

## Construcción de un dataframe conjunto para análisis de correlación

Ahora voy a unir ambas series de precios de cierre en un solo dataframe para poder comparar su comportamiento financiero de manera simultánea.

Esto permitirá calcular la correlación entre IBM y Walmart y analizar si existe relación estadística entre ambas series.

```python
df = pd.concat([ibm_close, wmt_close], axis=1)

df.columns = ["IBM", "Walmart"]

display(df.head())
```

## Cálculo de correlación entre ambas series

A continuación voy a calcular la matriz de correlación entre los precios de cierre de IBM y Walmart.

El coeficiente de correlación permite medir la fuerza y dirección de la relación lineal entre dos variables:

- cercano a 1 → correlación positiva fuerte
- cercano a 0 → poca correlación
- cercano a -1 → correlación negativa fuerte

```python
correlation = df.corr()
correlation
```

## Visualización de la matriz de correlación

Ahora voy a representar gráficamente la matriz de correlación mediante un mapa de calor para facilitar la interpretación visual de la relación entre ambas series financieras.

```python
plt.figure(figsize=(6,4))

sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Mapa de calor de correlación")

plt.show()
```

## Interpretación de correlación

Con base en el coeficiente de correlación obtenido, es posible evaluar si IBM y Walmart presentan cierto grado de relación estadística en el comportamiento de sus precios.

Una correlación positiva alta indicaría que ambas acciones tienden a moverse en una dirección similar en algunos periodos.

```python
print("Coeficiente de correlación:")
print(correlation)
```

## Descomposición de la serie de tiempo de IBM

Ahora voy a descomponer la serie temporal de IBM en sus componentes principales:

- tendencia
- estacionalidad
- residuales

Esto permite observar patrones ocultos dentro del comportamiento histórico de la serie.

```python
decompose_ibm = seasonal_decompose(ibm_close, model="additive", period=30)

decompose_ibm.plot()

plt.show()
```

## Correlograma de IBM

El correlograma permite observar la autocorrelación de la serie de tiempo a través de diferentes rezagos.

Esto ayuda a identificar dependencia temporal en los datos financieros.

```python
plot_acf(ibm_close, lags=50)

plt.show()
```

## Prueba de Dickey-Fuller para IBM

Ahora voy a aplicar la prueba estadística de Dickey-Fuller aumentada para evaluar si la serie de tiempo de IBM es estacionaria.

Interpretación general:

- p-value < 0.05 → serie estacionaria
- p-value > 0.05 → serie no estacionaria

```python
adf_result = adfuller(ibm_close.dropna())

print("ADF Statistic:", adf_result[0])
print("p-value:", adf_result[1])
print("Critical Values:")

for key, value in adf_result[4].items():
    print(f"{key}: {value}")
```

## Cálculo de promedios móviles

Ahora voy a calcular promedios móviles simples sobre la serie de IBM para suavizar el comportamiento de la serie y observar mejor la tendencia.

Se utilizarán:

- promedio móvil de 20 días
- promedio móvil de 50 días

```python
ibm_ma20 = ibm_close.rolling(window=20).mean()
ibm_ma50 = ibm_close.rolling(window=50).mean()

print("Promedios móviles calculados correctamente")
```

## Visualización de promedios móviles

En esta gráfica se compara:

- precio real de IBM
- promedio móvil de 20 días
- promedio móvil de 50 días

Esto ayuda a observar suavización y tendencia de la serie temporal.

```python
plt.figure(figsize=(12,6))

plt.plot(ibm_close, label="IBM Real")
plt.plot(ibm_ma20, label="MA 20")
plt.plot(ibm_ma50, label="MA 50")

plt.title("IBM con promedios móviles")
plt.xlabel("Fecha")
plt.ylabel("Precio")
plt.legend()

plt.show()
```

## Pronóstico simple mediante promedio móvil

Para generar un pronóstico simple del siguiente día de cotización, utilizaré el promedio de los últimos 20 días de IBM.

Este método no es un modelo avanzado, pero permite obtener una estimación sencilla basada en el comportamiento reciente de la serie.

```python
forecast_next = ibm_close.tail(20).mean()

print("Pronóstico simple para el siguiente día:")
print(forecast_next)
```
