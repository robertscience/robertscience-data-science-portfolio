## Objetivo del análisis
Se realiza un análisis exploratorio del dataset de ventas consolidadas con el fin de identificar patrones, validar calidad de datos, detectar anomalías y generar insights accionables para toma de decisiones.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10,6)
```

## 1. Carga del dataset

```python
path = '../robertscience_data/input/sales_consolidated_final.csv'

try:
    df = pd.read_csv(path)
    print('Dataset cargado correctamente')
except FileNotFoundError:
    print('No se encontró el archivo. Generando dataset sintético.')
    np.random.seed(42)
    df = pd.DataFrame({
        'DATE': pd.date_range('2024-01-01', periods=200),
        'TOTAL_VALUE_SALES': np.random.gamma(2, 500, 200),
        'QUANTITY': np.random.randint(1, 20, 200)
    })
```

## 2. Exploración inicial

```python
df.head()
df.info()
```

## 3. Calidad de datos

```python
print(df.isnull().sum())
print('Duplicados:', df.duplicated().sum())
```

## 4. Estadística descriptiva

```python
df.describe().T
```

## 5. Distribución de ventas

```python
plt.hist(df['TOTAL_VALUE_SALES'], bins=40, edgecolor='black')
plt.title('Distribución de ventas')
plt.show()
```

## 6. Outliers (IQR)

```python
q1 = df['TOTAL_VALUE_SALES'].quantile(0.25)
q3 = df['TOTAL_VALUE_SALES'].quantile(0.75)
iqr = q3 - q1

outliers = df[(df['TOTAL_VALUE_SALES'] < q1 - 1.5*iqr) | (df['TOTAL_VALUE_SALES'] > q3 + 1.5*iqr)]
print('Outliers:', len(outliers))
```

## 7. Feature engineering

```python
if 'QUANTITY' in df.columns:
    df['AVG_TICKET'] = df['TOTAL_VALUE_SALES'] / df['QUANTITY'].replace(0, np.nan)
```

## 8. Análisis temporal

```python
if 'DATE' in df.columns:
    df['DATE'] = pd.to_datetime(df['DATE'], errors='coerce')
    ts = df.groupby('DATE')['TOTAL_VALUE_SALES'].sum()
    ts.plot()
    plt.title('Ventas en el tiempo')
    plt.show()
```

## 9. Correlación

```python
sns.heatmap(df.corr(numeric_only=True), cmap='coolwarm')
plt.title('Correlación')
plt.show()
```

## 10. KPIs

```python
print('Total ventas:', df['TOTAL_VALUE_SALES'].sum())
print('Promedio:', df['TOTAL_VALUE_SALES'].mean())
print('Máximo:', df['TOTAL_VALUE_SALES'].max())
```

## 11. Conclusión

A partir del análisis exploratorio realizado, identifiqué un dataset consistente con variabilidad significativa en ventas.

Detecté valores atípicos relevantes que pueden afectar métricas agregadas si no se tratan correctamente.

El análisis temporal muestra fluctuaciones que sugieren patrones no lineales de demanda.

Considero que el dataset es adecuado para avanzar hacia modelado predictivo o segmentación, siempre que se complemente con ingeniería de variables y validación adicional.
