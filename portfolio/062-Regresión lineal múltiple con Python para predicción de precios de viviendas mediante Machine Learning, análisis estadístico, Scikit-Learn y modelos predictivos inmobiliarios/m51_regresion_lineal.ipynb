Después de cargar la información voy a revisar las dimensiones del dataset y visualizar las primeras filas para asegurarme de que los datos fueron importados correctamente.

```python
# Cargar dataset
df = pd.read_csv("kc_house_data.csv")

# Mostrar dimensiones del dataset
print("Dimensiones del dataset:")
print(df.shape)

# Mostrar primeras filas
df.head()
```

## Análisis de correlación entre variables

Ahora voy a analizar la correlación entre las variables numéricas del dataset.

Esto me ayudará a identificar qué variables tienen una relación más fuerte con el precio de las viviendas, ya que esas variables podrían ser importantes para el modelo de regresión lineal.

Para visualizar mejor esta información voy a generar un mapa de calor utilizando Seaborn.

## Relación entre metros cuadrados y precio

Después de revisar las correlaciones generales, ahora quiero visualizar directamente la relación entre los metros cuadrados habitables y el precio de las viviendas.

Esta gráfica me ayudará a identificar si existe una tendencia lineal entre ambas variables.

```python
# Relación entre sqft_living y price
plt.figure(figsize=(8,5))

sns.scatterplot(
    x=df["sqft_living"],
    y=df["price"]
)

plt.title("Relación entre metros cuadrados y precio")
plt.xlabel("Metros cuadrados habitables")
plt.ylabel("Precio")

plt.show()
```

## Selección de variables para el modelo

Ahora voy a seleccionar las variables que utilizaré para entrenar el modelo de regresión lineal.

En este caso utilizaré:
- metros cuadrados habitables
- cantidad de habitaciones
- cantidad de baños

Estas variables suelen tener una relación importante con el precio de una vivienda.

```python
# Variables independientes
X = df[["sqft_living", "bedrooms", "bathrooms"]]

# Variable dependiente
y = df["price"]

print("Variables preparadas correctamente")
```

## División de datos para entrenamiento y prueba

Ahora voy a dividir el dataset en datos de entrenamiento y datos de prueba.

Esto es importante porque me permitirá entrenar el modelo con una parte de la información y posteriormente evaluar qué tan bien funciona utilizando datos que el modelo no ha visto anteriormente.

En este caso utilizaré:
- 80% para entrenamiento
- 20% para prueba

```python
# División de datos
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Datos divididos correctamente")
print("Entrenamiento:", X_train.shape)
print("Prueba:", X_test.shape)
```

## Entrenamiento del modelo de regresión lineal

Ahora voy a crear y entrenar el modelo de regresión lineal utilizando los datos de entrenamiento.

El objetivo es que el modelo aprenda la relación entre las variables seleccionadas y el precio de las viviendas.

```python
# Crear modelo
model = LinearRegression()

# Entrenar modelo
model.fit(X_train, y_train)

print("Modelo entrenado correctamente")
```

## Generación de predicciones

Después de entrenar el modelo, ahora voy a generar predicciones utilizando los datos de prueba.

Estas predicciones me permitirán evaluar qué tan preciso es el modelo al estimar precios de viviendas.

```python
# Generar predicciones
y_pred = model.predict(X_test)

print("Predicciones generadas correctamente")
```

## Evaluación del modelo

Ahora voy a evaluar el rendimiento del modelo utilizando distintas métricas de machine learning.

Las métricas que voy a utilizar son:
- MAE (Error Absoluto Medio)
- MSE (Error Cuadrático Medio)
- R² Score

Estas métricas me ayudarán a medir qué tan preciso es el modelo realizando predicciones.

```python
# Calcular métricas
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

# Mostrar resultados
print("MAE:", mae)
print("MSE:", mse)
print("R2 Score:", r2)
```

## Comparación entre valores reales y predicciones

Ahora voy a comparar algunos valores reales contra las predicciones generadas por el modelo.

Esto me permitirá observar visualmente qué tan cercanas son las predicciones respecto a los precios reales de las viviendas.

```python
# Crear DataFrame comparativo
comparison = pd.DataFrame({
    "Precio Real": y_test,
    "Precio Predicho": y_pred
})

# Mostrar resultados
comparison.head(10)
```

## Visualización de predicciones del modelo

Ahora voy a generar una gráfica para visualizar la relación entre los valores reales y las predicciones realizadas por el modelo.

Si el modelo tiene buen rendimiento, los puntos deberían acercarse a una línea diagonal.

```python
# Gráfica de predicciones
plt.figure(figsize=(8,5))

plt.scatter(y_test, y_pred)

plt.xlabel("Valores Reales")
plt.ylabel("Predicciones")
plt.title("Valores reales vs predicciones")

plt.show()
```
