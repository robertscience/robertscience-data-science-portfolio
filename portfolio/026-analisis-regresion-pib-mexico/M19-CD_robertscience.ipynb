# Tarea M19-CD – Análisis de Regresión del PIB de México

## Resumen Ejecutivo
- **Objetivo:** Analizar la evolución histórica del PIB de México y generar un modelo predictivo.
- **Metodología:** Ajuste de modelo logístico mediante mínimos cuadrados, normalización de datos, evaluación de bondad de ajuste y pronóstico del PIB 2022.
- **Resultados esperados:** Pronóstico confiable y visualización profesional para entrega académica.

### Celda 1: Librerías y configuración
Se importan librerías esenciales y se define el estilo de las gráficas para un look profesional.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# Configuración de pandas y estilo de matplotlib
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
plt.style.use('seaborn-whitegrid')
```

### Celda 2: Carga del dataset
Se carga el archivo Excel `MexicoGDP.xlsx` y se visualizan las primeras filas para verificar la correcta importación de los datos.

```python
df = pd.read_excel("../MexicoGDP.xlsx")
df.head()
```

### Celda 3: Inspección del dataset
Revisamos dimensiones y tipos de datos para entender la estructura de la base de datos.

```python
print(df.shape)
df.info()
```

### Celda 4: Definición de variables X e Y
Asignamos X como años y Y como PIB, que serán utilizadas en el modelo de regresión.

```python
X = df.iloc[:,0]
Y = df.iloc[:,1]
X.head(), Y.head()
```

### Celda 5: Visualización histórica del PIB
Se realiza un gráfico de línea con puntos y sombreado para mostrar la evolución del PIB.

```python
plt.figure(figsize=(10,6))
plt.plot(X, Y, marker='o', color='darkblue', linewidth=2, label='PIB histórico')
plt.fill_between(X, Y, color='skyblue', alpha=0.3)
plt.title('Evolución histórica del PIB de México', fontsize=16, fontweight='bold')
plt.xlabel('Año', fontsize=12)
plt.ylabel('PIB (Millones USD)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```

### Celda 6: Definición del modelo logístico
Función logística con dos parámetros: beta1 y beta2, usada para ajuste de regresión no lineal.

```python
def modelo_logistico(x, beta1, beta2):
    return 1 / (1 + np.exp(-(beta1 + beta2*x)))
```

### Celda 7: Ajuste inicial sin normalización
Intentamos ajustar el modelo logístico con los datos originales. Puede presentar problemas de convergencia debido a la magnitud de los valores.

```python
parametros, _ = curve_fit(modelo_logistico, X, Y, maxfev=10000)
parametros
```

### Celda 8: Normalización de los datos
Se dividen los valores de cada columna entre su valor máximo, estabilizando la estimación de parámetros y facilitando el ajuste logístico.

```python
X_norm = X / X.max()
Y_norm = Y / Y.max()
X_norm.head(), Y_norm.head()
```

### Celda 9: Ajuste logístico con datos normalizados
Se realiza el ajuste nuevamente y se obtienen parámetros estables y confiables.

```python
parametros_norm, _ = curve_fit(modelo_logistico, X_norm, Y_norm, maxfev=10000)
parametros_norm
```

### Celda 10: Visualización del ajuste normalizado
Se muestran puntos de datos, línea de ajuste logístico y sombreado para resaltar la curva.

```python
plt.figure(figsize=(10,6))
plt.scatter(X_norm, Y_norm, color='navy', label='Datos Normalizados', s=50)
plt.plot(X_norm, modelo_logistico(X_norm, *parametros_norm), color='red', linewidth=2, label='Ajuste Logístico')
plt.fill_between(X_norm, modelo_logistico(X_norm, *parametros_norm), color='pink', alpha=0.2)
plt.title('Ajuste del Modelo Logístico al PIB Normalizado de México', fontsize=16, fontweight='bold')
plt.xlabel('Año (Normalizado)', fontsize=12)
plt.ylabel('PIB (Normalizado)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```

### Celda 11: Pronóstico PIB 2022
Se predice el PIB del 2022 usando el modelo logístico ajustado y se transforma a la escala original.

```python
anio_pred = 2022
anio_pred_norm = anio_pred / X.max()
pib_pred_norm = modelo_logistico(anio_pred_norm, *parametros_norm)
pib_pred = pib_pred_norm * Y.max()
print(f"Pronóstico del PIB para 2022: {pib_pred:.2f} millones USD")
```

### Celda 12: Comentarios finales y conclusiones
- La normalización permitió un ajuste estable.
- El modelo logístico captura la tendencia histórica.
- El pronóstico para 2022 está listo.
