El objetivo fue analizar la estructura de los datos y establecer una base confiable para procesos posteriores de limpieza, transformación e integración.

## 1. Importación de librerías

Utilicé las librerías necesarias para manipulación y análisis de información.

```python
import pandas as pd
from pathlib import Path
```

## 2. Configuración de rutas

Definí la ruta de trabajo tomando como referencia la estructura del repositorio.

La carpeta `data/raw` representa la ubicación donde se almacenan las fuentes originales utilizadas durante el desarrollo.

```python
BASE_DIR = Path.cwd().parent

DATA_PATH = BASE_DIR / 'data' / 'raw'

DATA_PATH
```

## 3. Carga de fuentes

Realicé la carga inicial de las fuentes comerciales utilizadas para el análisis.

Las fuentes corresponden a dimensiones descriptivas y una tabla principal de hechos de ventas.

```python
df_calendar = pd.read_excel(DATA_PATH / 'DIM_CALENDAR.xlsx')

df_category = pd.read_csv(DATA_PATH / 'DIM_CATEGORY.csv')

df_product = pd.read_excel(DATA_PATH / 'DIM_PRODUCT.xlsx')

df_segment = pd.read_excel(DATA_PATH / 'DIM_SEGMENT.xlsx')

df_sales = pd.read_csv(DATA_PATH / 'FACT_SALES.csv')
```

## 4. Organización de datasets

Organicé las fuentes dentro de un diccionario para facilitar su análisis y validación.

```python
datasets = {
    'DIM_CALENDAR': df_calendar,
    'DIM_CATEGORY': df_category,
    'DIM_PRODUCT': df_product,
    'DIM_SEGMENT': df_segment,
    'FACT_SALES': df_sales
}

datasets.keys()
```

## 5. Exploración inicial

Realicé una revisión estructural para identificar dimensiones, columnas disponibles y características generales de cada fuente.

```python
for nombre, dataframe in datasets.items():
    
    print('=' * 70)
    print(f'Dataset: {nombre}')
    print('=' * 70)
    
    print('Dimensiones:')
    print(dataframe.shape)
    
    print('\nColumnas:')
    print(dataframe.columns.tolist())
    
    print('\n')
```

## 6. Revisión de tipos de datos

Analicé los tipos de variables disponibles para identificar la estructura de cada fuente.

```python
for nombre, dataframe in datasets.items():
    
    print('=' * 70)
    print(nombre)
    print('=' * 70)
    
    print(dataframe.dtypes)
    print('\n')
```

## 7. Validación de valores nulos

Realicé una revisión inicial de calidad para identificar posibles valores faltantes dentro de las fuentes.

```python
for nombre, dataframe in datasets.items():
    
    print('=' * 70)
    print(f'Valores nulos: {nombre}')
    print('=' * 70)
    
    print(dataframe.isnull().sum())
    print('\n')
```
